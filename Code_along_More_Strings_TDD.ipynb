# count_vowels(s)

The function count_vowels(s) takes a string s and returns the number of vowels (a, e, i, o, u) in the string.

```python
def count_vowels(s):
    '''
    This function takes a string and returns
    the number of vowels (a, e, i, o, u) in s
    (case-insensitive)
    '''
    count= 0
    for letter in s:
        if letter.lower() in "aeiou":
            count += 1
    return count
```

## Tests

Below you will see tests for the count_vowels function

```python
def test_count_vowels():
  assert count_vowels("aeiou") == 5, 'Failed: count_vowels("aeiou") == 5'
  assert count_vowels("aaaeeiiiioooouu") == 15, 'Failed: count_vowels("aaaeeiiiioooouu") == 15'
  assert count_vowels("a b c d e f ") == 2, 'Failed: count_vowels("a b c d e f ") == 2'
  assert count_vowels("1234567") == 0, 'Failed: count_vowels("1234567") == 0'
  assert count_vowels("%!&#^*#") == 0, 'Failed: count_vowels("%!&#^*#") == 0'
  assert count_vowels(" A E I O U") == 5, 'Failed: count_vowels(" A E I O U") == 5'
  print("All count_vowels tests passed")
```

```python
test_count_vowels()
```

# get_handle(s)

Write a function get_handle(email) that extracts and returns the handle part of an email address (everything before the '@' symbol).

```python
def get_handle(s):
  '''
      This function extracts and returns the handle part of an email address (everything before the '@' symbol)
  '''

  # One method is to use .find() as you can pass in a string and it'll tell you what index it's at
  index = s.find("@")
  # Return slices (uses []) up to but not including index that contains '@'
  # -1 in this context doesn't mean end of the string. It's a special error code the find method returns
  # If there is no '@' then index = -1 and will return "" else index = spot where '@' is located and return splices before it
  return s[0:index] if index != -1 else ""
```

## Tests for get_handle(s)

```python
def test_get_handle():
  assert get_handle("suland@greenriver.edu") == "suland", 'Failed: get_handle("suland@greenriver.edu") == "suland"'
  assert get_handle("my_email@somewhere.com") == "my_email", 'Failed: get_handle("my_email@somewhere.com") == "my_email"'
  assert get_handle("this is not an email") == "", 'Failed: get_handle("this is not an email") == ""'
  print("All get_handle tests passed")
```

```python
test_get_handle()
```

# is_all_digits(s)

Write a function is_all_digits(s) that checks if a string s contains only digits. Return True if it does, and False otherwise.

```python
def is_all_digits(s):
  '''
      This function checks if a string s contains only digits. Return True if it does, and False otherwise
  '''

  # There is a built-in function that does the work for us
  return s.isdigit()
```

## Tests for is_all_digits(s)

```python
def test_is_all_digits():
  assert is_all_digits("012345678") == True, "Failed: is_all_digits('012345678') == True"
  assert is_all_digits("hello5") == False, "Failed: is_all_digits('hello5') == False"
  assert is_all_digits("!@abc") == False, "Failed: is_all_digits('!@abc') == False"
  print("test_is_all_digits passed")
```

```python
test_is_all_digits()
```

##compare_to(str1, str2)

Write a function that returns an integer value of 1 if str1 comes after str2 and an integer value of -1 if str1 comes before str2 lexiographically. If str1 is equal to str2 return 0.  In other words, this function will let you know if str1 comes before or after str2 alphabetically based on the integer returned.  

```python
def compare_to(str1, str2):
  '''
      Best practice - Always add a doc string

      This function returns an integer value of 1 if str1 comes after str2 and an integer value of -1 if str1 comes before str2 lexiographically. If str1 is equal to str2 return 0
  '''
  # You can compare strings with relational operators in python
  # A comes before B, B comes before b
  if str1 > str2:
    return 1
  elif str1 < str2:
    return -1
  else:
    return 0
```

## Tests compare_to(str1, str2)

```python
def test_compare_to():
  assert compare_to("Apple", "Banana") == -1, "Failed: compare_to('Apple', 'Banana') == -1"
  assert compare_to("Banana", "Apple") == 1, "Failed: compare_to('Banana', 'Apple') == 1"
  assert compare_to("Banana", "Banana") == 0, "Failed: compare_to('Banana', 'Banana') == 0"
  assert compare_to("banana", "Banana") == 1, "Failed: compare_to('banana', 'Banana') == 1"
  print("test_compared_to all tests passed")
```

```python
test_compare_to()
```
