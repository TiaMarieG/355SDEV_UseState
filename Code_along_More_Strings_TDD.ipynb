# count_vowels(s)

The function count_vowels(s) takes a string s and returns the number of vowels (a, e, i, o, u) in the string.

```python
def count_vowels(s):
    '''
    This function takes a string and returns
    the number of vowels (a, e, i, o, u) in s
    (case-insensitive)
    '''
    count= 0
    for letter in s:
        if letter.lower() in "aeiou":
            count += 1
    return count
```

## Tests

Below you will see tests for the count_vowels function

```python
def test_count_vowels():
  assert count_vowels("aeiou") == 5, 'Failed: count_vowels("aeiou") == 5'
  assert count_vowels("aaaeeiiiioooouu") == 15, 'Failed: count_vowels("aaaeeiiiioooouu") == 15'
  assert count_vowels("a b c d e f ") == 2, 'Failed: count_vowels("a b c d e f ") == 2'
  assert count_vowels("1234567") == 0, 'Failed: count_vowels("1234567") == 0'
  assert count_vowels("%!&#^*#") == 0, 'Failed: count_vowels("%!&#^*#") == 0'
  assert count_vowels(" A E I O U") == 5, 'Failed: count_vowels(" A E I O U") == 5'
  print("All count_vowels tests passed")
```

```python
test_count_vowels()
```
